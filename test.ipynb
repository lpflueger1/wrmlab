```python
# Random Jupyter Notebook demo code
# Generates random data, analyzes it, and plots results

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# Create random dataset
data = pd.DataFrame({
    "x": np.arange(1, 101),
    "y": np.random.normal(loc=50, scale=15, size=100),
    "category": np.random.choice(["A", "B", "C"], size=100)
})

# Show first rows
print("Sample Data:")
display(data.head())

# Basic statistics
print("\nStatistics:")
display(data.describe())

# Grouped averages
group_avg = data.groupby("category")["y"].mean()
print("\nAverage by Category:")
display(group_avg)

# Plot
plt.figure(figsize=(10, 5))
for cat in data["category"].unique():
    subset = data[data["category"] == cat]
    plt.scatter(subset["x"], subset["y"], label=f"Category {cat}")

plt.title("Random Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.legend()
plt.grid(True)

plt.show()
```
